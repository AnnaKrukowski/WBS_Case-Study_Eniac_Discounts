```python
# Option 3
# Filtering only 2017 to have one complete year and including where discount_total > 0
all_2017 = all[(all['created_date'].dt.year == 2017) & (all['discount_total'] > 0)].copy()


# New column 'month' based on 'created_date'
all_2017.loc[:, 'month'] = all_2017['created_date'].dt.month

all_2017.loc[:, 'discount_rel'] = ((all_2017['discount_total']/all_2017['price_total'])*100).round(2)

# New DataFrame for the plot: monthly sales sum of 'unit_price_total'
monthly_discount_2017 = all_2017.groupby('month').agg({'discount_rel': 'mean'}).reset_index()

monthly_discount_2017
```

```python
merged_sales_discount=monthly_sales_2017.merge(monthly_discount_2017, how='left',on='month')
merged_sales_discount
```

```python
# Creating the plot
fig, ax1 = plt.subplots(figsize=(12, 6))

# Plotting the sales trend on the primary y-axis
sns.lineplot(data=merged_sales_discount, x='month', y='unit_price_total', ax=ax1, label='Sales in mio. €', color='steelblue')
ax1.set_xlabel('Month',fontsize=14,)
ax1.set_ylabel('Sales in mio. €', fontsize=14, color='steelblue')
ax1.tick_params(axis='y', labelcolor='steelblue')

# Creating the secondary y-axis
ax2 = ax1.twinx()
sns.lineplot(data=merged_sales_discount, x='month', y='discount_rel', ax=ax2, color='lightslategray', linestyle='--', label='Mean Discount (%)')
ax2.set_ylabel('Mean Discount (%)', fontsize=14, color='lightslategray')
ax2.tick_params(axis='y', labelcolor='lightslategray')
ax2.set_ylim(0, merged_sales_discount['discount_rel'].max() + 5)

# Adding the title
plt.title('Monthly Sales Trend and Mean Discount Rates for 2017',fontsize=14)

# Adding legends
lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines + lines2, labels + labels2, loc='lower right', bbox_to_anchor=(0.9,0.1),fontsize=12)

plt.show()
```

### Analyze special dates

```python
# Creating new df with the necessary columns
df = all_2017[['created_date', 'unit_price_total','month']]

# Filtering the df for the desired months (January, July, November, December)
df_filtered = df[df['created_date'].dt.month.isin([1, 7, 11, 12])]

# Grouping by date and calculate the sum of unit_price_total
df_grouped = df_filtered.groupby(df_filtered['created_date'].dt.date)['unit_price_total'].sum().reset_index()

# Transforming Datatypes
df_grouped['created_date']=pd.to_datetime(df_grouped['created_date'], errors='coerce')
df_grouped['month'] = df_grouped['created_date'].dt.month

# Finding the top 5 dates for each month
top_dates = pd.concat([df_grouped[df_grouped['created_date'].dt.month == month].nlargest(5, 'unit_price_total') for month in [1, 7, 11, 12]])

# Sorting the top dates chronologically for the plot
top_dates_sorted = top_dates.sort_values(by='created_date').reset_index(drop=True)

# Connecting/Mapping months to colors
month_colors = {1: 'yellow', 7: 'orange', 11: 'red', 12: 'saddlebrown'}
palette = top_dates_sorted['month'].map(month_colors)

# Creating the bar plot
plt.figure(figsize=(12, 6))
bars = sns.barplot(data=top_dates_sorted, x='created_date', y='unit_price_total',hue='month', palette=month_colors)
plt.xlabel('Date')
plt.ylabel('Sales')
plt.title('Top 5 Dates for January, July, November, and December (2017) with the highest Sales')
plt.xticks(rotation=45)
plt.grid(True)
#plt.legend(title='Month', labels=['January', 'July', 'November', 'December']) # no colors in legend

# Setting the legend manually
handles, labels = bars.get_legend_handles_labels()
labels = ['January', 'July', 'November', 'December']
plt.legend(handles, labels, title='Month')

plt.show()
```

### Relationship between recommended price for a product (products.price) and the price that products actually sold for (orderlines.unit_price)

```python
def discount_category(row):
    if (row['price'] - row['unit_price']) > 0:
        return 'Discounted Price'
    elif row['price'] == row['unit_price']:
        return 'Recommended Price'
    else:
        return 'Above Retail'

merged_df=orderlines.merge(products,how='left',on='sku') # merge

merged_df['discount'] = merged_df.apply(discount_category, axis=1)

sns.relplot(data=merged_df,
            x="price",
            y="unit_price",
            hue="discount",
            );

plt.xlabel('Recommended Price')
plt.ylabel('Price at which it was sold')
plt.title('Relationship between recommended price and the price at which it was sold');
```
